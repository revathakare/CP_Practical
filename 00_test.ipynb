```python
#student grade A-90 to 100 marks, B-80 to 89 marks, C-70 to 79 marks, D-60 to 69 marks, E-50 to 59 marks, F-below 50 marks
#if elif else format
marks = int(input("Enter the student's marks: "))

if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
elif marks >= 50:
    grade = "E"
else:
    grade = "F"
print(f"The student's grade is: {grade}")
```
